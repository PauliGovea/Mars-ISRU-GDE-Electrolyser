```python
J_COL = "Current Density_mean"
ERR   = "sem"          # "sem" | "std" | "iqr" | None

CENTRE = "median" if ERR == "iqr" else "mean"
CO = f"FE(CO)_{CENTRE}"
H2 = f"FE(H2)_{CENTRE}"

if ERR in ("sem", "std"):
    CO_ERR, H2_ERR = f"FE(CO)_{ERR}", f"FE(H2)_{ERR}"
elif ERR == "iqr":
    CO_ERR = ["FE(CO)_err_low", "FE(CO)_err_high"]
    H2_ERR = ["FE(H2)_err_low", "FE(H2)_err_high"]
else:
    CO_ERR = H2_ERR = None

def _yerr(g, spec):
    if spec is None:
        return None
    if isinstance(spec, list):
        return np.vstack([g[spec[0]].values, g[spec[1]].values])
    return g[spec].values
```

```python
color_H2, color_CO = "#CCCCCC", "#0072B2"

def _bars(ax, summary, co, h2, co_err, h2_err):
    x = np.arange(len(summary)) # Draw one stacked panel. co/h2 = height columns, *_err = yerr arrays
    ax.bar(x, summary[co], 0.6, color=color_CO, edgecolor="black", lw=1, label="CO")
    ax.bar(x, summary[h2], 0.6, bottom=summary[co],
           color=color_H2, edgecolor="black", lw=1, label=r"H$_2$")
    ax.errorbar(x, summary[co], yerr=co_err,
                fmt="none", ecolor="black", capsize=3, lw=1)
    ax.errorbar(x, summary[co] + summary[h2], yerr=h2_err,
                fmt="none", ecolor="red", capsize=3, lw=1)
    ax.axhline(100, ls=":", color="grey", lw=1)
    ax.set_xticks(x)
    ax.set_xticklabels(summary["Current Density_mean"].round(0).astype(int))
    ax.grid(axis="y", ls="--", alpha=0.4)


def get(conc, pH): # Rows for one condition, ordered by current density
    s = combined_df[(combined_df["Concentration"] == conc) &
                    (combined_df["pH"] == pH)].sort_values("Current Density_mean")
    if s.empty:
        raise KeyError(f"no rows for {conc}, {pH}")
    return s
```

## Mean and SEM

```python
def plot_mean(conc, pH, ax=None):
    s = get(conc, pH)
    if ax is None:
        fig, ax = plt.subplots(figsize=(8, 7))
    _bars(ax, s,
          co="FE(CO)_mean", h2="FE(H2)_mean",
          co_err=s["FE(CO)_sem"].values,
          h2_err=s["FE(H2)_sem"].values)
    return ax
```

## Median and IQR

```python
def plot_median(conc, pH, ax=None):
    s = get(conc, pH)
    if ax is None:
        fig, ax = plt.subplots(figsize=(8, 7))
    _bars(ax, s,
          co="FE(CO)_median", h2="FE(H2)_median",
          co_err=np.vstack([s["FE(CO)_err_low"], s["FE(CO)_err_high"]]),
          h2_err=np.vstack([s["FE(H2)_err_low"], s["FE(H2)_err_high"]]))
    return ax
```

```python
concs = ["0.3M", "0.6M", "0.9M"]
pHs   = ["pH1", "pH2", "pH3"]

def grid(plot_fn, title=None, ymax=105, save=None):
    fig, axes = plt.subplots(3, 3, figsize=(14, 14), sharey=True)
    for i, conc in enumerate(concs):
        for j, pH in enumerate(pHs):
            ax = axes[i, j]
            plot_fn(conc, pH, ax=ax)
            ax.set_ylim(0, ymax)
            if i == 0:
                ax.set_title(pH.replace("pH", "pH "), fontsize=15)
            if i == 2:
                ax.set_xlabel(r"- J [A m$^{-2}$]", fontsize=13)
            if j == 0:
                ax.set_ylabel(rf"{conc} Na$_2$SO$_4$" "\n" "Faradaic efficiency (%)",
                              fontsize=13)
    axes[0, 0].legend(fontsize=11, loc="upper right")

    if title:
        fig.suptitle(title, fontsize=18, y=0.995)
    plt.tight_layout()
    if title:
        plt.subplots_adjust(top=0.94)      # room for the suptitle
    if save:
        plt.savefig(save, dpi=300, bbox_inches="tight")
    plt.show()
```

```python
from scipy.interpolate import griddata
o3M_pH1['Depth'] = 0.3
o6M_pH1['Depth'] = 0.6
o9M_pH1['Depth'] = 0.9
# l2M_pH1['Depth'] = 1.2
# l5M_pH1['Depth'] = 1.5
# l8M_pH1['Depth'] = 1.8

# Combine the dataframes
# combined_pH1 = pd.concat([o3M_pH1, o6M_pH1, o9M_pH1, l2M_pH1, l5M_pH1, l8M_pH1])
combined_pH1 = pd.concat([o3M_pH1, o6M_pH1, o9M_pH1])
combined_pH1.fillna(0, inplace=True)

# Create the grid for the contour plot
current_density_values = np.unique(combined_pH1['Current Density_mean'])
depth_values = np.array([0.3, 0.6, 0.9])
X1, Y1 = np.meshgrid(current_density_values, depth_values)

# Prepare data for interpolation
points = combined_pH1[['Current Density_mean', 'Depth']].values
values = combined_pH1['FE(CO)_mean'].values

# Interpolate missing values
Z1 = griddata(points, values, (X1, Y1), method='linear')

# Clip values to range 0-100
Z1 = np.clip(Z1, 0, 100)

# Plot the contour
plt.figure(figsize=(10, 6))
contour = plt.contourf(X1, Y1, Z1, levels=20, cmap='inferno', vmin=0, vmax=100)
cbar = plt.colorbar(contour, label='FE(CO)', ticks=np.linspace(0, 100, 11))
plt.xlabel(r'J [mA cm$^{-2}$]')
plt.ylabel('Concentration [M] Na$_2$SO$_4$')
plt.title('Contour Plot of pH1 Na$_2$SO$_4$')
plt.yticks(ticks=[0.3, 0.6, 0.9], labels=['0.3M', '0.6M', '0.9M'])  # Set y-axis ticks
plt.xlim(20,3000)
# plt.savefig("All_pH1_mean_contour.png")
plt.show()
```
